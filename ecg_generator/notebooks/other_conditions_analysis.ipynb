```python
import matplotlib.pyplot as plt
from ecg_model import generate_ecg
```

```python
conditions = {
    "Normal": 70,
    "Tachycardia": 120,
    "Bradycardia": 40
}
```

```python
plt.figure(figsize=(10,8))

for i, (name, hr) in enumerate(conditions.items(),1):

    t, signal = generate_ecg(6, hr)

    plt.subplot(3,1,i)
    plt.plot(t, signal,color="blue")

    plt.title(name)
    plt.grid()

plt.tight_layout()
plt.show()
```
